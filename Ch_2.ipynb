Function

```python
def area(lenght):
    area = lenght * lenght
    return area

b = area(5)
print('area = %.2f' %b)
```

Lambda

```python
cal = lambda x: x * 2

a = cal(5)
b = cal(10)
print(a)
print(b)
```

List

```python
data = [1, 2, 3, 4, 5]
print(data)
type(data)
```

```python
data.append(6) #add 6 to the end of the list
print(data)
```

```python
avg = sum(data) / len(data)
print('Average = %.2f' %avg)
```
